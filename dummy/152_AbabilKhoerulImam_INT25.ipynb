```python
import random

data = random.sample(range(1, 100), 10)
print("Data awal:", data)
```

```python
def linear_search(arr, target):
    steps = 0
    for i in range(len(arr)):
        steps += 1
        if arr[i] == target:
            return i, steps
    return -1, steps

target = data[5]  
index, steps_ls = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Jumlah langkah Linear Search (acak):", steps_ls)
```

```python
def bubble_sort(arr):
    arr = arr.copy()
    n = len(arr)
    steps = 0
    
    for i in range(n-1):
        for j in range(n-1-i):
            steps += 1
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                
    return arr, steps

bubble_sorted, steps_bubble = bubble_sort(data)

print("Hasil Bubble Sort:", bubble_sorted)
print("Jumlah langkah Bubble Sort:", steps_bubble)
```

```python
def selection_sort(arr):
    arr = arr.copy()
    n = len(arr)
    steps = 0
    
    for i in range(n-1):
        min_index = i
        for j in range(i+1, n):
            steps += 1
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
        
    return arr, steps

selection_sorted, steps_selection = selection_sort(data)

print("Hasil Selection Sort:", selection_sorted)
print("Jumlah langkah Selection Sort:", steps_selection)
```

```python
def insertion_sort(arr):
    arr = arr.copy()
    steps = 0
    
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        
        while j >= 0 and arr[j] > key:
            steps += 1
            arr[j+1] = arr[j]
            j -= 1
        
        arr[j+1] = key
        
    return arr, steps

insertion_sorted, steps_insertion = insertion_sort(data)

print("Hasil Insertion Sort:", insertion_sorted)
print("Jumlah langkah Insertion Sort:", steps_insertion)
```

```python
target_sorted = bubble_sorted[5]  
index_sorted, steps_ls_sorted = linear_search(bubble_sorted, target_sorted)

print("Target:", target_sorted)
print("Index ditemukan:", index_sorted)
print("Jumlah langkah Linear Search (terurut):", steps_ls_sorted)
```
